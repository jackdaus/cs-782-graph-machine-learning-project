```python
import plotly.express as px
import numpy as np

# Generate sample data
t = np.linspace(0, 2 * np.pi, 100)
x = np.sin(t)
y = np.cos(t)
z = t

# Create a 3D scatter plot
fig = px.scatter_3d(x=x, y=y, z=z)

# Show the plot
fig.show()
```
