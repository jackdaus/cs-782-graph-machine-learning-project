# Basic Pipeline

## 01 Create the database and set up directories

```python
import pathlib

# Define an output path
output_path = pathlib.Path("./output")
# Create the output path
output_path.mkdir(exist_ok=True)
# Define database path
database_path = output_path / "database.db"


# Defined the image directory
image_dir = pathlib.Path("./input/lego")
```

## 02 Run pipeline

```python
import pycolmap

# Extract Features
pycolmap.extract_features(database_path, image_dir, camera_model='OPENCV')
# Feature Matching
pycolmap.match_exhaustive(database_path)
# Bundle Adjustment
# reconstruction = pycolmap.incremental_mapping(database_path, image_dir, output_path)
```

```python
maps = pycolmap.incremental_mapping(database_path, image_dir, output_path)
```
